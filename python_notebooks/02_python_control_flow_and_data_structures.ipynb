# Python Control Flow and Data Structures

This notebook focuses on two central elements of Python programming:

1. **Control flow**: deciding which instructions are executed and how often.
2. **Data structures**: organizing and manipulating collections of values.

## Learning objectives

By the end of this notebook, you should be able to use:

- `if`, `elif`, and `else`;
- logical operators;
- `for` and `while` loops;
- `break` and `continue`;
- lists, tuples, sets, and dictionaries;
- slicing;
- list comprehensions;
- iteration patterns for simple data processing.

## 1. Why control flow matters

A program does not always execute every instruction in the same way.

We often want to express logic such as:

- **if** something happens, do one thing;
- **otherwise**, do something else;
- **repeat** an operation for several items;
- **continue** until a condition becomes false.

These mechanisms define the program's control flow.

## 2. `if`

````python
temperature = 38.4

if temperature >= 38:
    print("Fever")
````

## 3. `if` / `else`

````python
temperature = 37.1

if temperature >= 38:
    print("Fever")
else:
    print("No fever")
````

## 4. `if` / `elif` / `else`

````python
temperature = 37.7

if temperature >= 39:
    print("High fever")
elif temperature >= 38:
    print("Fever")
elif temperature >= 37.5:
    print("Slightly elevated temperature")
else:
    print("Normal temperature")
````

## 5. Combining conditions

````python
age = 72
has_diabetes = True

if age >= 65 and has_diabetes:
    print("Both conditions are satisfied")
````

````python
temperature = 36.9
has_symptoms = True

if temperature >= 38 or has_symptoms:
    print("Further assessment may be useful")
````

## 6. Nested conditions

You can place an `if` statement inside another one.

However, deeply nested logic can become difficult to read.

````python
age = 70
temperature = 38.5

if age >= 65:
    if temperature >= 38:
        print("Older patient with fever")
````

Often the same logic can be written more clearly:

````python
if age >= 65 and temperature >= 38:
    print("Older patient with fever")
````

# Loops

## 7. `for` loops

A `for` loop iterates over the elements of a collection.

````python
temperatures = [36.5, 37.2, 38.0, 39.1]

for temperature in temperatures:
    print(temperature)
````

````python
for temperature in temperatures:
    if temperature >= 38:
        print(f"{temperature} -> fever")
    else:
        print(f"{temperature} -> no fever")
````

## 8. `range()`

`range()` generates a sequence of integers.

````python
for i in range(5):
    print(i)
````

````python
for i in range(2, 6):
    print(i)
````

````python
for i in range(0, 10, 2):
    print(i)
````

````python
# range backwards
for i in range(10, 0, -1):
    print(i)
````

## 9. `enumerate()`

Use `enumerate()` when you need both an index and a value.

````python
patients = ["Anna", "Marco", "Elena"]

for index, patient in enumerate(patients):
    print(index, patient)
````

## 10. `zip()`

`zip()` lets us iterate over multiple sequences in parallel.

````python
patients = ["Anna", "Marco", "Elena"]
temperatures = [36.8, 38.2, 37.1]

for patient, temperature in zip(patients, temperatures):
    print(f"{patient}: {temperature} °C")
````

````python
patients = ["Anna", "Marco", "Elena"]
temperatures = [36.8, 38.2, 37.1]
ages = [12, 7, 5]

for patient, temperature, age in zip(patients, temperatures, ages):
    print(f"{age} yo {patient}: {temperature} °C")
````

## 11. `while` loops

A `while` loop continues as long as its condition is true.

````python
count = 0

while count < 5:
    print(count)
    count += 1
````

Be careful with infinite loops: the loop condition must eventually become false.

## 12. `break`

`break` stops the loop immediately.

````python
values = [4, 7, 12, 3, 9]

for value in values:
    if value > 10:
        print("Found:", value)
        break
````

## 13. `continue`

`continue` skips the rest of the current iteration.

````python
values = [10, -2, 7, -1, 4]

for value in values:
    if value < 0:
        continue

    print(value)
````

# Data structures

## 14. Lists

Lists are:

- ordered;
- mutable;
- able to contain duplicate values.

````python
temperatures = [36.8, 37.1, 38.2, 36.9]

print(temperatures)
print(type(temperatures))
````

### Indexing

````python
print(temperatures[0])
print(temperatures[2])
print(temperatures[-1])
````

### Modifying a list

````python
temperatures = [36.8, 37.1, 38.2]

temperatures.append(39.0)
print(temperatures)

temperatures[0] = 36.6
print(temperatures)
````

````python
values = [3, 1, 4]

values.append(2)
print("append:", values)

values.insert(1, 10)
print("insert:", values)

values.remove(4)
print("remove:", values)

last_value = values.pop()
print("pop:", last_value)
print(values)
````

## 15. Slicing

Slicing selects part of a sequence.

```python
sequence[start:stop]
```

The `stop` index is excluded.

````python
values = [10, 20, 30, 40, 50, 60]

print(values[1:4])
print(values[:3])
print(values[3:])
print(values[::2])
print(values[3:])
print(values[1:5:2])
````

## 16. Tuples

Tuples are ordered but **immutable**.

````python
coordinates = (45.65, 13.77)

print(coordinates[0])
print(coordinates[1])
````

### Tuple unpacking

````python
latitude, longitude = coordinates

print(latitude)
print(longitude)
````

## 17. Sets

A `set` contains unique elements.

Sets are useful for:

- removing duplicates;
- membership tests;
- set operations.

````python
departments = {"Cardiology", "Neurology", "Cardiology", "Oncology"}

print(departments)
````

````python
group_a = {"A", "B", "C"}
group_b = {"B", "C", "D"}

print("Union:", group_a | group_b)
print("Intersection:", group_a & group_b)
print("Difference:", group_a - group_b)
````

## 18. Dictionaries

Dictionaries store **key-value pairs**.

````python
patient = {
    "name": "Anna",
    "age": 52,
    "temperature": 37.8,
    "smoker": False
}

print(patient)
````

### Accessing values

````python
print(patient["name"])
print(patient["temperature"])
````

The method `get()` is useful when a key may not exist.

````python
print(patient.get("blood_pressure"))
print(patient.get("blood_pressure", "Not available"))
````

### Modifying a dictionary

````python
patient["temperature"] = 38.1
patient["blood_pressure"] = 125

print(patient)
````

### Iterating over dictionaries

````python
for key in patient:
    print(key)
````

````python
for value in patient.values():
    print(value)
````

````python
for key, value in patient.items():
    print(key, "->", value)
````

````python
temperatures = {
    "Anna": 37.2,
    "Marco": 38.4,
    "Luca": 36.8
}

fever = {
    name: temp
    for name, temp in temperatures.items()
    if temp >= 38
}

print(fever)
````

## 19. Nested data structures

A common pattern is a **list of dictionaries**.

````python
patients = [
    {"name": "Anna", "age": 52, "temperature": 37.8},
    {"name": "Marco", "age": 67, "temperature": 38.4},
    {"name": "Elena", "age": 45, "temperature": 36.9}
]

print(patients[0])
print(patients[0]["name"])
````

**Dictionary of dictionaries**

````python
patients_dict = {
    "P001": {
        "name": "Anna",
        "age": 54,
        "temperature": 37.2
    },
    "P002": {
        "name": "Marco",
        "age": 67,
        "temperature": 38.4
    }
}

print(patients_dict["P002"]["temperature"])
````

````python
for patient in patients:
    print(patient["name"], patient["temperature"])
````

## 20. Filtering data with a loop

````python
patients_with_fever = []

for patient in patients:
    if patient["temperature"] >= 38:
        patients_with_fever.append(patient)

print(patients_with_fever)
````

## 21. List comprehensions

List comprehensions provide a compact way to create lists.

General pattern:

```python
new_list = [expression for item in collection]
```

````python
temperatures = [36.5, 37.0, 38.2, 39.1]

temperatures_fahrenheit = [
    temperature * 9 / 5 + 32
    for temperature in temperatures
]

print(temperatures_fahrenheit)
````

You can also add a condition.

````python
fever_temperatures = [
    temperature
    for temperature in temperatures
    if temperature >= 38
]

print(fever_temperatures)
````

````python
import time

n = 100_000_000

start = time.perf_counter()

squares = []
for x in range(n):
    squares.append(x * x)

end = time.perf_counter()

print("For loop:", end - start)


start = time.perf_counter()

squares = [x * x for x in range(n)]

end = time.perf_counter()

print("List comprehension:", end - start)
````

Use list comprehensions for simple transformations.

If the logic becomes complicated, a regular `for` loop is usually easier to read.

# Common programming patterns

## 22. Accumulator

````python
temperatures = [36.5, 37.0, 38.2, 39.1]

total = 0

for temperature in temperatures:
    total += temperature

mean_temperature = total / len(temperatures)

print(mean_temperature)
````

Python already provides `sum()`:

````python
mean_temperature = sum(temperatures) / len(temperatures)
print(mean_temperature)
````

## 23. Counter

````python
number_with_fever = 0

for temperature in temperatures:
    if temperature >= 38:
        number_with_fever += 1

print(number_with_fever)
````

## 24. Search

````python
target = 38.2
found = False

for temperature in temperatures:
    if temperature == target:
        found = True
        break

print(found)
````

The membership operator `in` can often express this more directly:

````python
print(target in temperatures)
````

# Mini case study

We have a small collection of patients.

We want to:

1. display patient names;
2. identify patients with fever;
3. compute the mean age;
4. build a list containing only the names of patients with fever.

````python
patients = [
    {"name": "Anna", "age": 52, "temperature": 37.8},
    {"name": "Marco", "age": 67, "temperature": 38.4},
    {"name": "Elena", "age": 45, "temperature": 36.9},
    {"name": "Luca", "age": 73, "temperature": 38.9},
]

for patient in patients:
    print(patient["name"])
````

````python
for patient in patients:
    if patient["temperature"] >= 38:
        print(patient["name"], "has fever")
````

````python
ages = []

for patient in patients:
    ages.append(patient["age"])

mean_age = sum(ages) / len(ages)

print(mean_age)
````

````python
patients_with_fever = [
    patient["name"]
    for patient in patients
    if patient["temperature"] >= 38
]

print(patients_with_fever)
````
