# Functions and Good Practices

Functions are one of the most important tools for writing clear, reusable, and maintainable Python programs.

A function groups a sequence of instructions under a name. Instead of rewriting the same logic every time we need it, we can define it once and call it whenever necessary.

## Learning objectives

By the end of this notebook, you should be able to:

- define and call functions;
- understand the difference between parameters and arguments;
- return values from functions;
- use positional, keyword, and default arguments;
- return multiple values;

- structure code into small reusable units;
- apply basic Python style principles.

## 1. Why functions?

Imagine that we need to perform the same operation several times.

Without a function, we may end up copying and pasting the same code. This creates several problems:

- the program becomes longer;
- repeated code is harder to read;
- if the logic changes, we need to modify it in many places;
- inconsistencies and bugs become more likely.

A **function** lets us give a name to a reusable computation.

The general syntax is:

```python
def function_name():
    # instructions
```

The indented block belongs to the function.

Defining a function does **not** execute it. We execute it by calling its name followed by parentheses.

````python
def greet():
    print("Hello!")

greet()
````

Here:

- `def` tells Python that we are defining a function;
- `greet` is the function name;
- `()` contains the parameters, if any;
- the indented line is the function body;
- `greet()` calls the function.

## 2. Parameters and arguments

Functions become much more useful when they can receive input values.

A **parameter** is the name used in the function definition.

An **argument** is the actual value supplied when the function is called.

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Anna")
greet("Marco")
````

In the definition:

```python
def greet(name):
```

`name` is a **parameter**.

In the call:

```python
greet("Anna")
```

`"Anna"` is an **argument**.

The same function can therefore operate on different inputs without changing its code.

## 3. Returning values

A function can compute a result and send it back to the caller using `return`.

This is usually preferable to printing the result directly when the function is meant to be reused elsewhere in the program.

````python
def square(x):
    return x ** 2

result = square(5)
print(result)
````

When Python reaches:

```python
return x ** 2
```

the function ends and the computed value is returned to the caller.

The returned value can then be:

- stored in a variable;
- passed to another function;
- used inside an expression;
- printed later.

### `print()` and `return` are different

`print()` displays information on the screen.

`return` sends a value back to the part of the program that called the function.

A function that does not explicitly return a value returns `None`.

````python
def square_print(x):
    print(x ** 2)

def square_return(x):
    return x ** 2

a = square_print(4)
b = square_return(4)

print("a:", a)
print("b:", b)
````

The first function prints `16`, but the variable `a` receives `None`.

The second function returns `16`, so the value can be stored in `b` and reused later.

For reusable computational functions, returning values is usually more flexible than printing them.

## 4. Multiple parameters

A function can receive several inputs.

````python
def bmi(weight_kg, height_m):
    return weight_kg / height_m**2

print(bmi(70, 1.75))
````

Here the function has two parameters:

- `weight_kg`
- `height_m`

When we call:

```python
bmi(70, 1.75)
```

Python associates the arguments with the parameters according to their position:

```text
weight_kg = 70
height_m  = 1.75
```

These are called **positional arguments**.

## 5. Default arguments

Sometimes a parameter has a sensible value that should be used most of the time.

We can specify a **default value** directly in the function definition.

````python
def convert_temperature(celsius, decimals=1):
    fahrenheit = celsius * 9 / 5 + 32
    return round(fahrenheit, decimals)

print(convert_temperature(37))
print(convert_temperature(37, decimals=3))
````

If `decimals` is not provided, Python uses the default value `1`.

Therefore:

```python
convert_temperature(37)
```

is equivalent to:

```python
convert_temperature(37, 1)
```

The caller can still override the default whenever needed.

Parameters without default values must normally appear before parameters with default values.

## 6. Keyword arguments

Instead of relying only on the position of arguments, we can explicitly state the parameter name.

These are called **keyword arguments**.

````python
print(
    convert_temperature(
        celsius=36.8,
        decimals=2
    )
)
````

Keyword arguments can improve readability, especially when:

- a function has many parameters;
- several parameters have similar types;
- optional parameters are being changed;
- the meaning of a value is not obvious from its position alone.

For example:

```python
convert_temperature(36.8, 2)
```

works, but:

```python
convert_temperature(celsius=36.8, decimals=2)
```

makes the meaning of each value explicit.

## 7. Returning multiple values

A function can appear to return more than one value.

````python
def min_max(values):
    return min(values), max(values)

minimum, maximum = min_max([3, 8, 1, 5])

print("Minimum:", minimum)
print("Maximum:", maximum)
````

Internally, Python is actually returning a **tuple**.

We can verify this:

````python
result = min_max([3, 8, 1, 5])

print(result)
print(type(result))
````

The expression:

```python
return min(values), max(values)
```

creates a tuple such as:

```python
(1, 8)
```

Python then allows us to **unpack** that tuple directly:

```python
minimum, maximum = min_max(values)
```

This is convenient when a computation naturally produces a small number of related results.

## 8. Variable numbers of arguments: `*args`

Sometimes we do not know in advance exactly how many positional arguments a function will receive.

Using `*args`, Python collects all additional positional arguments into a tuple.

````python
def average(*values):
    return sum(values) / len(values)

print(average(10, 20))
print(average(10, 20, 30, 40))
````

Inside the function, `values` is simply a tuple.

````python
def show_arguments(*values):
    print(values)
    print(type(values))

show_arguments(1, 2, 3)
````

`*args` is useful when the number of inputs is naturally variable.

The name `args` is only a convention. For example, writing:

```python
def average(*values):
```

is often clearer because `values` describes what the arguments represent.

## 9. Variable keyword arguments: `**kwargs`

`**kwargs` performs a similar role for keyword arguments.

Python collects them into a dictionary.

````python
def show_patient(**info):
    print(info)
    print(type(info))

show_patient(
    name="Anna",
    age=45,
    smoker=False
)
````

The variable `info` contains:

```python
{
    "name": "Anna",
    "age": 45,
    "smoker": False
}
```

This technique is useful for flexible interfaces, although it should not replace clearly named parameters when the expected inputs are known in advance.

## 10. Functions are objects

In Python, functions are themselves objects.

This means that a function can be:

- assigned to a variable;
- passed to another function;
- stored inside a data structure;
- returned by another function.

We do not need advanced functional programming to make use of this idea.

````python
def square(x):
    return x ** 2

operation = square

print(operation(5))
````

Notice the difference between:

```python
square
```

and:

```python
square(5)
```

`square` refers to the function object itself.

`square(5)` calls the function and evaluates its result.

A function can therefore receive another function as an argument.

````python
def apply_operation(x, operation):
    return operation(x)

def square(x):
    return x ** 2

def double(x):
    return 2 * x

print(apply_operation(4, square))
print(apply_operation(4, double))
````

This pattern appears frequently in Python libraries.

For example, functions such as `sorted()` can receive another function through their `key` parameter to decide how objects should be ordered.

## 11. Local scope

Variables created inside a function normally belong to the function's **local scope**.

````python
def compute_total():
    total = 10 + 20
    return total

print(compute_total())

# `total` only exists inside the function.
````

The variable `total` is created when the function runs and is local to `compute_total()`.

Trying to access it outside the function would produce an error:

```python
print(total)
```

This separation is useful because different functions can use the same local variable names without interfering with each other.

In general, prefer passing information explicitly through parameters and return values instead of relying heavily on global variables.

## 12. Docstrings

A **docstring** is text placed at the beginning of a function that explains what the function does.

It is written using triple quotes.

A good docstring is especially helpful when:

- someone else will use the function;
- the code will be reused later;
- the function is part of a larger project or library.

````python
def mean(values):
    """Return the arithmetic mean of a non-empty sequence."""
    return sum(values) / len(values)

print(mean([10, 20, 30]))
print(mean.__doc__)
````

A more detailed docstring can describe:

- the purpose of the function;
- its parameters;
- optional/default values;
- the returned result.

````python
def classify_temperature(temperature, threshold=38.0):
    """
    Classify a temperature measurement.

    Parameters
    ----------
    temperature : float
        Temperature in degrees Celsius.
    threshold : float, default=38.0
        Fever threshold.

    Returns
    -------
    str
        "fever" or "no fever".
    """
    if temperature >= threshold:
        return "fever"

    return "no fever"
````

````python
print(classify_temperature.__doc__)
````

````python
help(classify_temperature)
````

There are several accepted docstring styles. The exact format is less important than being clear and consistent.

## 13. Type hints

Type hints document the expected types of function inputs and outputs.

````python
def add(a: float, b: float) -> float:
    return a + b

print(add(2.5, 3.0))
````

In:

```python
def add(a: float, b: float) -> float:
```

we are communicating that:

- `a` is expected to be a `float`;
- `b` is expected to be a `float`;
- the function is expected to return a `float`.

Type hints improve readability and can be used by editors, IDEs, and static-analysis tools.

However, Python normally does **not** enforce them at runtime.

````python
# Python may still execute this:
print(add("Hello ", "world"))
````

So type hints are mainly documentation and tooling support unless additional validation mechanisms are used.

## 14. A common pitfall: mutable default arguments

This is an important Python behavior.

Consider the following function:

````python
def add_patient(name, patients=[]):
    patients.append(name)
    return patients

print(add_patient("Anna"))
print(add_patient("Marco"))
````

You might expect the second call to return:

```python
["Marco"]
```

but it returns:

```python
["Anna", "Marco"]
```

Why?

Default arguments are evaluated **once, when the function is defined**, not every time the function is called.

Therefore the same list object is reused across calls.

This matters for mutable objects such as:

- lists;
- dictionaries;
- sets.

The usual solution is to use `None` as the default and create a new object inside the function.

````python
def add_patient(name, patients=None):
    if patients is None:
        patients = []

    patients.append(name)
    return patients

print(add_patient("Anna"))
print(add_patient("Marco"))
````

Now each call creates its own list unless a list is explicitly supplied by the caller.

This pattern is very common in Python:

```python
def function_name(values=None):
    if values is None:
        values = []
```

# Good practices

The following principles are not strict rules, but they make functions easier to understand, test, reuse, and maintain.

## 15. One clear responsibility

A function should ideally perform one well-defined task.

A function that mixes unrelated responsibilities is more difficult to test and modify.

For example, this function:

- displays information;
- performs classification;
- computes a new value.

````python
def process_patient(name, age, temperature):
    print(name)
    print(age)

    if temperature >= 38:
        print("fever")
    else:
        print("no fever")

    return age + 1
````

A clearer design is to separate those responsibilities into small functions.

````python
def has_fever(temperature, threshold=38.0):
    return temperature >= threshold


def next_age(age):
    return age + 1


def patient_summary(name, age, temperature):
    status = "fever" if has_fever(temperature) else "no fever"
    return f"{name}, age {age}, status: {status}"


print(patient_summary("Anna", 40, 38.2))
````

Each function now has a clear purpose.

This makes the code easier to:

- understand;
- test independently;
- reuse;
- modify later.

## 16. DRY: Don't Repeat Yourself

**DRY** means avoiding unnecessary duplication of logic.

If exactly the same computation appears repeatedly, it may be useful to extract it into a function.

````python
def percentage(part, total):
    return 100 * part / total

print(percentage(25, 100))
print(percentage(18, 60))
````

Instead of rewriting:

```python
100 * part / total
```

everywhere, the function gives the operation a meaningful name and keeps the implementation in one place.

However, not every repeated line needs its own function. The goal is clarity, not creating functions for every tiny operation.

## 17. Guard clauses

A **guard clause** handles a special or invalid case early and returns immediately.

This often avoids deeply nested `if` statements.

````python
def safe_divide(a, b):
    if b == 0:
        return None

    return a / b

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

The special case is handled first:

```python
if b == 0:
    return None
```

After that, the main computation can remain simple.

Guard clauses are particularly useful for:

- invalid input;
- empty collections;
- missing information;
- boundary cases.

## 18. Use meaningful names

Good function and variable names communicate intent.

Prefer:

```python
mean_temperature
number_of_patients
calculate_risk_score
```

over:

```python
x
n2
calc
```

unless the short name is obvious from the context.

A useful rule is:

> Names should describe what a value represents or what a function does.

## 19. Keep functions reasonably short

A function that becomes very long may be doing too many things.

Splitting complex logic into smaller functions can improve:

- readability;
- testing;
- reuse;
- maintenance.

There is no universal maximum number of lines.

The more useful question is:

> Can I describe what this function does in one short sentence?

If the answer requires several unrelated actions, the function may benefit from being split.

## 20. Avoid hidden side effects

A **side effect** is a change that happens outside the returned result of a function.

Examples include:

- modifying a global variable;
- modifying an input list in place;
- writing to a file;
- printing to the screen.

Side effects are not always bad, but hidden or unnecessary side effects can make code harder to understand.

````python
# Clear and predictable
def normalize(value, minimum, maximum):
    return (value - minimum) / (maximum - minimum)

print(normalize(7, 0, 10))
````

This function is easy to reason about:

- all required information comes through parameters;
- the function computes a result;
- the result is returned;
- no external state is modified.

Such functions are often easier to test and reuse.

# Possible solution

````python
def summarize_measurements(values: list[float]) -> dict:
    """Return simple summary statistics for a non-empty list of values."""

    if len(values) == 0:
        return {}

    return {
        "min": min(values),
        "max": max(values),
        "mean": sum(values) / len(values),
        "count": len(values),
    }


print(summarize_measurements([10, 12, 15, 9]))
````

- Python can return multiple values by returning a tuple.
- `*args` collects positional arguments into a tuple.
- `**kwargs` collects keyword arguments into a dictionary.
