## Challenge — Make `__getitem__` smarter

Modify `__getitem__` so that both of these work:

```python
catalog[0]
catalog["Laptop"]
```

If the key is an integer or a slice, use normal list indexing.

If the key is a string, search for a product with that name.

Optional: raise a `KeyError` when no product with that name exists.

````python
class ProductCatalog:
    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)

    def __iter__(self):
        return iter(self.products)

    def __getitem__(self, key):
        if isinstance(key, str):
            for product in self.products:
                if product["name"] == key:
                    return product
            raise KeyError(f"Product '{key}' not found")
        return self.products[key]

    def __contains__(self, name):
        return any(product["name"] == name for product in self.products)

    def cheaper_than(self, price):
        return [p for p in self.products if p["price"] < price]

    def by_category(self, category):
        return [p for p in self.products if p["category"] == category]
````

## Task 8 — Define the neural network class

Use:

````python
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(2, 4)
        self.relu = nn.ReLU()
        self.output = nn.Linear(4, 1)

    def forward(self, x):
        x = self.hidden(x)
        x = self.relu(x)
        return self.output(x)
````

## Task 9 — Create a training class

````python
class Trainer:
    def __init__(self, model, learning_rate=0.05):
        self.model = model
        self.loss_fn = nn.BCEWithLogitsLoss()
        self.optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
````

## Task 10 — Add a `train` method

For every epoch:

1. compute the model predictions,
2. compute the loss,
3. reset old gradients,
4. call `loss.backward()`,
5. update the parameters with `optimizer.step()`.

Print the loss every 100 epochs.

````python
# Add this method inside Trainer:

def train(self, X, y, epochs=500):
    for epoch in range(epochs):
        predictions = self.model(X)
        loss = self.loss_fn(predictions, y)
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

        if epoch % 100 == 0:
            print("Epoch:", epoch, "Loss:", loss.item())
````

## Complete neural-network reference solution

This version puts all methods inside their classes and can be run directly.


````python
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(2, 4)
        self.relu = nn.ReLU()
        self.output = nn.Linear(4, 1)

    def forward(self, x):
        x = self.hidden(x)
        x = self.relu(x)
        return self.output(x)

    def predict(self, x):
        self.eval()
        with torch.no_grad():
            logits = self(x)
            probabilities = torch.sigmoid(logits)
            return (probabilities >= 0.5).int()


class Trainer:
    def __init__(self, model, learning_rate=0.05):
        self.model = model
        self.loss_fn = nn.BCEWithLogitsLoss()
        self.optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

    def train(self, X, y, epochs=500):
        self.model.train()
        for epoch in range(epochs):
            predictions = self.model(X)
            loss = self.loss_fn(predictions, y)

            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()

            if epoch % 100 == 0:
                print("Epoch:", epoch, "Loss:", loss.item())
````

````python
model = SimpleNN()
trainer = Trainer(model)
trainer.train(X, y)

print(model.predict(torch.tensor([[4.2, 3.8]])))
print(model.predict(torch.tensor([[1.2, 1.1]])))
````

````python
from torch.utils.data import DataLoader, TensorDataset

class Trainer:
    def __init__(self, model, learning_rate=0.05):
        self.model = model
        self.loss_fn = nn.BCEWithLogitsLoss()
        self.optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

    def train(self, X, y, epochs=500, batch_size=32):

        dataset = TensorDataset(X, y)
        dataloader = DataLoader(
            dataset,
            batch_size=batch_size,
            shuffle=True
        )

        for epoch in range(epochs):
            self.model.train()
            total_loss = 0.0

            for X_batch, y_batch in dataloader:

                # Forward pass
                predictions = self.model(X_batch)
                loss = self.loss_fn(predictions, y_batch)

                # Backpropagation
                self.optimizer.zero_grad()
                loss.backward()
                self.optimizer.step()

                total_loss += loss.item() * len(X_batch)

            # Average loss over all training samples
            avg_loss = total_loss / len(dataset)

            if epoch % 100 == 0:
                print("Epoch:", epoch, "Loss:", avg_loss)
````

````python

class MyDataLoader:
    def __init__(self, X, y, batch_size=32, shuffle=True):
        self.X = X
        self.y = y
        self.batch_size = batch_size
        self.shuffle = shuffle

    def __iter__(self):
        n = len(self.X)

        if self.shuffle:
            indices = torch.randperm(n)
        else:
            indices = torch.arange(n)

        for start in range(0, n, self.batch_size):
            batch_indices = indices[start:start + self.batch_size]

            X_batch = self.X[batch_indices]
            y_batch = self.y[batch_indices]

            yield X_batch, y_batch  # yield: returns one batch at the time. no need to create a list with all batches

    def __len__(self):
        return (len(self.X) + self.batch_size - 1) // self.batch_size
````
